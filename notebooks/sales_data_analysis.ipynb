# Sales Data Analysis - Beginner Project

## Introduction

This notebook explores a simple sales dataset to practice data cleaning, analysis, and visualization using Python. 

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
%matplotlib inline
```

## 2. Load the Dataset

```python
df = pd.read_csv("../data/sales_data.csv")
df.head()
```

## 3. Basic Information

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

## 4. Data Cleaning

```python
# Check for missing values
df.isnull().sum()
```

```python
# Check for duplicates
df.duplicated().sum()
```

```python
# Calculate total sales amount
df["Total Sales"] = df["Quantity"] * df["Unit Price"]
df.head()
```

## 5. Exploratory Data Analysis (EDA)

```python
# Sales by Category
df.groupby("Category")["Total Sales"].sum().sort_values(ascending=False)
```

```python
# Sales by Region
df.groupby("Region")["Total Sales"].sum()
```

## 6. Visualizations

```python
# Bar chart - Total Sales by Category
plt.figure(figsize=(8,5))
df.groupby("Category")["Total Sales"].sum().plot(kind="bar", color=["skyblue", "lightgreen", "salmon"])
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Histogram - Distribution of Unit Price
plt.figure(figsize=(8,5))
plt.hist(df["Unit Price"], bins=8, color="purple", alpha=0.7)
plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price ($)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(6,5))
numeric_cols = df.select_dtypes(include=["int64", "float64"])
sns.heatmap(numeric_cols.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 7. Key Insights

- **Category Performance**: Different categories show varying total sales amounts
- **Regional Spread**: Sales distribution across North, South, East, West regions
- **Pricing**: Unit prices range from low-cost stationery to higher-priced electronics
- **Quantity**: Most orders have small quantities, typical for a small business dataset
- **Clean Data**: No missing values or duplicates found in this sample

## Conclusion

This simple analysis demonstrates the power of basic EDA. Even with a small dataset, we can extract useful insights with just a few lines of code!
